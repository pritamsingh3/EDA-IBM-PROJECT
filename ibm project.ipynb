```python
# 2. Which feature has the highest standard deviation?
df.select_dtypes(include='number').std().sort_values(ascending=False)
```

```python
# # 3. How can missing values affect Machine Learning models?


# Training errors 🚫
# Many algorithms (e.g., Linear Regression, SVM, Logistic Regression) cannot handle missing values directly and may throw errors.
# Reduced accuracy 📉
# Missing data can hide important patterns, leading to poor predictions.
# Biased results ⚠️
# If missing values are concentrated in certain groups, the model may learn incorrect relationships.
# Loss of data 🗑️
# Removing rows with missing values can reduce the dataset size and valuable information.
# Incorrect statistics 📊
# Missing values can distort calculations such as mean, median, correlation, and standard deviation.
```

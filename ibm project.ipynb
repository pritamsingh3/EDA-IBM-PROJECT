Part 3: Data Cleaning


```python

# Task 3.1: Missing Value Treatment

# For numerical columns:
# Apply:
# Mean Imputation
nums_cols = df.select_dtypes(include='number').columns
for i in nums_cols:
    df[i] = df[i].fillna(df[i].mean()) 


```

```python
# Median Imputation
nums_col = df.select_dtypes(include='number').columns
for i in nums_col:
    df[i] = df[i].fillna(df[i].median())
```

```python
# For categorical columns:
# Apply:

# Mode Imputation
cat_cols = df.select_dtypes(exclude='number').columns
for i in cat_cols:
    df[i] = df[i].fillna(df[i].mode()[0])
# Unknown Category Imputation

cat_cols = df.select_dtypes(exclude='number').columns

for col in cat_cols:
    df[col] = df[col].fillna('Unknown')
```

```python
# Questions

# 1. Which imputation method worked best?
# Median imputation worked best for numerical columns because it is less affected by outliers, 

# while Mode or Unknown Category imputation worked best for categorical columns.

# # 2. Why is median preferred for skewed distributions?
# Median is preferred for skewed distributions because it is resistant to outliers and extreme values. 
# Unlike the mean, the median is not heavily influenced by unusually large or small observations, 
# making it a more reliable measure of central tendency for skewed data.
```

Part 4: Univariate Analysis

Univariate Analysis is the statistical analysis of a single variable to understand 

its distribution, central tendency, variability, and overall characteristics. 

It is the first step of Exploratory Data Analysis (EDA).

```python
# Task 4.1: Numerical Features Analysis

# For each numerical feature:
# Calculate:
nums_col = df.select_dtypes(include='number')
# Mean
pd.DataFrame( { 'Mean' : df.select_dtypes(include='number').mean() } )
# Median
pd.DataFrame( { 'Median' : df.select_dtypes(include='number').median() } )
# Mode
pd.DataFrame( { 'Mode' : df.select_dtypes(include='number').mode().iloc[0] } )
# Variance
nums_col.var()
# Standard Deviation
nums_col.std()
# Minimum
nums_col.min()
# Maximum
nums_col.max()
# Range
nums_col.max()-nums_col.min()
# Quartiles
nums_col.quantile([0.25,0.50,0.75,1])

```

```python
# Create:
# Histogram
plt.figure(figsize=(6,4))
plt.hist(df['Age'],bins=10)
plt.title("Histogram of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()

```

```python
# KDE Plot
plt.Figure(figsize=(6,4))
sns.kdeplot(df['Age'],fill=True)
plt.title("KDE Plot of Age")
plt.show()
```

```python
nums_col = df.select_dtypes(include='number').columns
for i in nums_col:
    plt.Figure(figsize=(15,4))

    # histogram
    plt.subplot(1,3,1)
    plt.hist(df[i],bins=10)
    plt.title(f'histogram - {i}')

    # KDE
    plt.subplot(1,3,2)
    sns.kdeplot(df[i], fill=True)
    plt.title(f'KDE - {i}')

    # Boxplot
    plt.subplot(1,3,3)
    sns.boxplot(x=df[i])
    plt.title(f'Boxplot - {i}')

    plt.tight_layout()
    plt.show()
```
